## Performance by Algorithm

The benchmark results can be grouped by algorithm to observe how execution time and memory usage change with input size.

```python
for algorithm in df["Algorithm"].unique():
    print("\n", algorithm)
    display(df[df["Algorithm"] == algorithm])
```

## Execution Time Comparison

The following graph shows how execution time changes with input size for the different algorithms and program structures.

```python
plt.figure(figsize=(10, 6))

for algorithm in df["Algorithm"].unique():
    subset = df[df["Algorithm"] == algorithm]
    plt.plot(
        subset["Input Size"],
        subset["Execution Time (seconds)"],
        marker="o",
        label=algorithm
    )

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Algorithm Execution Time Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

## Search Algorithm Analysis

This section compares Linear Search and Binary Search using the experimental benchmark results.

Linear Search has a theoretical time complexity of O(n), while Binary Search has a theoretical time complexity of O(log n) for a sorted dataset.

The benchmark measures how their execution time changes as the input size increases.

```python
search_df = df[
    df["Algorithm"].isin(["Linear Search", "Binary Search"])
]

search_df
```

### Search Execution Time Comparison

```python
plt.figure(figsize=(9, 5))

for algorithm in ["Linear Search", "Binary Search"]:
    subset = search_df[search_df["Algorithm"] == algorithm]

    plt.plot(
        subset["Input Size"],
        subset["Execution Time (seconds)"],
        marker="o",
        label=algorithm
    )

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Linear Search vs Binary Search")
plt.legend()
plt.grid(True)
plt.show()
```

## Loop-Based Program Analysis

This section compares Single Loop Traversal and Nested Loop Traversal.

Single Loop Traversal has O(n) time complexity because the loop processes the input once.

Nested Loop Traversal has O(n²) time complexity because the inner loop executes for every iteration of the outer loop.

```python
loop_df = df[
    df["Algorithm"].isin(["Single Loop", "Nested Loop"])
]

loop_df
```

```python
plt.figure(figsize=(9, 5))

for algorithm in ["Single Loop", "Nested Loop"]:
    subset = loop_df[loop_df["Algorithm"] == algorithm]

    plt.plot(
        subset["Input Size"],
        subset["Execution Time (seconds)"],
        marker="o",
        label=algorithm
    )

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Single Loop vs Nested Loop")
plt.legend()
plt.grid(True)
plt.show()
```

## Factorial Algorithm Analysis

This section compares Recursive Factorial and Iterative Factorial.

Both implementations have O(n) time complexity.

Recursive Factorial requires O(n) auxiliary space because of the recursive call stack, while Iterative Factorial requires O(1) auxiliary space.

```python
factorial_df = df[
    df["Algorithm"].isin([
        "Recursive Factorial",
        "Iterative Factorial"
    ])
]

factorial_df
```

```python
plt.figure(figsize=(9, 5))

for algorithm in ["Recursive Factorial", "Iterative Factorial"]:
    subset = factorial_df[factorial_df["Algorithm"] == algorithm]

    plt.plot(
        subset["Input Size"],
        subset["Execution Time (seconds)"],
        marker="o",
        label=algorithm
    )

plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Recursive vs Iterative Factorial")
plt.legend()
plt.grid(True)
plt.show()
```

## Memory Usage Analysis

Memory usage is another important factor when evaluating algorithm performance.

The benchmarking engine records the peak memory used during each execution using Python's `tracemalloc` module.

This section compares the memory consumption of the implemented algorithms and program structures.

```python
plt.figure(figsize=(10, 6))

for algorithm in df["Algorithm"].unique():
    subset = df[df["Algorithm"] == algorithm]

    plt.plot(
        subset["Input Size"],
        subset["Memory (KB)"],
        marker="o",
        label=algorithm
    )

plt.xlabel("Input Size")
plt.ylabel("Peak Memory (KB)")
plt.title("Memory Usage Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

## Theoretical Complexity

The theoretical complexities of the implemented algorithms are:

| Algorithm | Time Complexity | Space Complexity |
|---|---|---|
| Linear Search | O(n) | O(1) |
| Binary Search | O(log n) | O(1) |
| Single Loop | O(n) | O(1) |
| Nested Loop | O(n²) | O(1) |
| Recursive Factorial | O(n) | O(n) |
| Iterative Factorial | O(n) | O(1) |

The experimental measurements are compared with these theoretical expectations to understand how algorithmic complexity affects practical performance.

## Experimental Observations

### Linear Search vs Binary Search

Linear Search examines elements sequentially and therefore its execution time can increase as the input size grows.

Binary Search repeatedly reduces the search space by half and therefore has logarithmic time complexity for sorted data.

### Single Loop vs Nested Loop

Single Loop Traversal performs one pass through the input and has O(n) time complexity.

Nested Loop Traversal performs approximately n² operations and therefore has O(n²) time complexity.

The benchmark results show a much faster increase in execution time for the nested loop as the input size increases.

### Recursive Factorial vs Iterative Factorial

Both factorial implementations have O(n) time complexity.

Recursive Factorial requires additional call-stack space because each recursive call remains active until the base case is reached.

Iterative Factorial uses a loop and constant auxiliary space.

The experimental results allow the theoretical time and space complexity to be compared with actual measurements.
